```python
from langchain.agents import create_agent
from langchain.messages import HumanMessage
from dotenv import load_dotenv
from pprint import pprint
from langchain.tools import tool
from tavily import TavilyClient
from typing import Dict, Any

from pydantic import BaseModel

load_dotenv()
```

```python
# Simple code to show tool creation in LangChain.
@tool
def square_root(num: float) -> float:
    """Returns the square root of a number."""
    return num**0.5
```

```python
# Invoking the tool directly with a single parameter input.
square_root.invoke({"num": 16})
```

```python
agent = create_agent(
    model="anthropic:claude-haiku-4-5",
    system_prompt="You are a mathmatician that solves problems based on the tools you have access to.",
    tools=[square_root],
)
```

```python
response = agent.invoke(
    {
        "messages": [
            HumanMessage(content="What's square root of 16?"),
        ],
    }
)
```

```python
pprint(response['messages'][-1].content)
```

```python
tavily_client = TavilyClient()
```

```python
# Create a web search tool
@tool
def web_search(query:str) -> Dict[str, Any]:
    """Searches the web for a given query and returns the results."""
    return tavily_client.search(query=query)
```

```python
web_search.invoke({"query": "Who is the current prime minister of India?"})
```

```python
web_search_agent = create_agent(
    model="anthropic:claude-sonnet-5",
    system_prompt="You are a helpful assistant that can search the web for information.",
    tools=[web_search],
)
```

```python
response = web_search_agent.invoke(
    {
        "messages": [
            HumanMessage(content="Who is the current prime minister of India?"),
        ],
    }
)
```

```python
pprint(response['messages'][-1].content)
```
