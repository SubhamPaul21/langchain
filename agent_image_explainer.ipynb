```python
from langchain.agents import create_agent
from langchain.messages import HumanMessage
from dotenv import load_dotenv
from pprint import pprint

load_dotenv()
```

```python
from ipywidgets import FileUpload
from IPython.display import display
```

```python
uploader = FileUpload(accept=".png", multiple=False)
display(uploader)
```

```python
print(uploader.value)
```

```python
import base64

uploaded_file = uploader.value[0]

content_mv = uploaded_file["content"]

image_bytes = bytes(content_mv)

image_base64 = base64.b64encode(image_bytes).decode("utf-8")
```

```python
multimodal_question = HumanMessage(
    content=[
        {
            "type": "text",
            "text": "Explain the following image in detail. Additionally, can you guess which anime this image is from?",
        },
        {"type": "image", "base64": image_base64, "mime_type": "image/png"},
    ]
)
```

```python
agent = create_agent(
    model="anthropic:claude-haiku-4-5"
)
```

```python
response = agent.invoke(
    {
        "messages": [multimodal_question],
    },
)
```

```python
print(response["messages"][-1].content)
```
