```python
from dotenv import load_dotenv
from tavily import TavilyClient
from typing import Dict, Any
from requests import get
from fastmcp import FastMCP
```

```python
mcp = FastMCP("mcp_basic_server")

tavily_client = TavilyClient()
```

```python
# Tool for searching the web
@mcp.tool()
def search_web(query: str) -> Dict[str, Any]:
    """
    Search the web for a given query and return the results.
    """
    results = tavily_client.search(query)
    return results
```

```python
# Resources for provided access to langchain-ai repo files
@mcp.resource("github://langchain-ai/langchain-mcp-adapters/blob/main/README.md")
def github_file():
    """
    Fetch the content of a file from the langchain-ai GitHub repository.
    """
    url = "https://raw.githubusercontent.com/langchain-ai/langchain-mcp-adapters/blob/main/README.md"
    response = get(url)
    if response.status_code == 200:
        return response.text
    else:
        return f"Failed to fetch file. Status code: {response.status_code}"
```

```python
# Prompt Template
@mcp.prompt()
def prompt_template(name: str, age: int) -> str:
    """
    Generate a prompt based on the provided name and age.
    """
    return f"Hello {name}, you are {age} years old."
```
